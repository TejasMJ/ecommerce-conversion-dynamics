#### Importing the libraries

```python
import pandas as pd
import numpy as np
```

#### Loading the prepared dataset

```python
df = pd.read_csv(
    '../data/processed/prepared_data.csv'
)
```

```python
df.head()
```

#### Creating a Copy of the Dataset

```python
# Creating a copy of the prepared dataset for feature engineering.

df_engineered = df.copy()
```

#### Creating Total Pages Viewed
The dataset separates browsing activity into three page categories:

- Administrative
- Informational
- ProductRelated

For conversion analysis, combining them gives us a broader measure of overall browsing activity during a session.


```python
# Creating a feature representing the total number of pages viewed during a session.

df_engineered['Total_Pages_Viewed'] = (
    df_engineered['Administrative']
    + df_engineered['Informational']
    + df_engineered['ProductRelated']
)
```

```python
# Displaying the newly created feature.

df_engineered[
    [
        'Administrative',
        'Informational',
        'ProductRelated',
        'Total_Pages_Viewed'
    ]
].head()
```

#### Creating Total Browsing Duration

The dataset contains browsing duration across the same three page categories. We can combine them into one session-level measure of total browsing time.

```python
# Creating a feature representing the total browsing duration during a session.

df_engineered['Total_Browsing_Duration'] = (
    df_engineered['Administrative_Duration']
    + df_engineered['Informational_Duration']
    + df_engineered['ProductRelated_Duration']
)
```

```python
# Displaying the newly created feature.

df_engineered[
    [
        'Administrative_Duration',
        'Informational_Duration',
        'ProductRelated_Duration',
        'Total_Browsing_Duration'
    ]
].head()
```

#### Creating Product Page Share

Product Page Share is created to represent the proportion of total pages viewed that are product-related. This feature helps capture the extent to which a session is focused on product browsing.

```python
# Creating a feature representing the proportion of pages that are product-related.

df_engineered['Product_Page_Share'] = (
    df_engineered['ProductRelated']
    / df_engineered['Total_Pages_Viewed'].replace(0, np.nan)
)
```

```python
# Displaying the newly created feature.

df_engineered[
    [
        'ProductRelated',
        'Total_Pages_Viewed',
        'Product_Page_Share'
    ]
].head()
```

#### Creating Session Engagement Ratio

We already have:

- Total_Pages_Viewed
- Total_Browsing_Duration

We can derive a useful session-level measure showing how much browsing time was spent per page viewed.

```python
# Creating a feature representing the average browsing duration per page viewed.

df_engineered['Avg_Duration_Per_Page'] = (
    df_engineered['Total_Browsing_Duration']
    / df_engineered['Total_Pages_Viewed'].replace(0, np.nan)
)
```

```python
# Displaying the newly created feature.

df_engineered[
    [
        'Total_Browsing_Duration',
        'Total_Pages_Viewed',
        'Avg_Duration_Per_Page'
    ]
].head()
```

#### Creating Product Browsing Duration Share

This complements Product_Page_Share. Instead of looking at the number of pages, this measures how much of the total browsing time is spent on product-related pages.

```python
# Creating a feature representing the proportion of browsing time spent on product-related pages.

df_engineered['Product_Duration_Share'] = (
    df_engineered['ProductRelated_Duration']
    / df_engineered['Total_Browsing_Duration'].replace(0, np.nan)
)
```

```python
# Displaying the newly created feature.

df_engineered[
    [
        'ProductRelated_Duration',
        'Total_Browsing_Duration',
        'Product_Duration_Share'
    ]
].head()
```

#### Reviewing Engineered Features

```python
# Checking the engineered features created so far.

engineered_columns = [
    'Total_Pages_Viewed',
    'Total_Browsing_Duration',
    'Product_Page_Share',
    'Avg_Duration_Per_Page',
    'Product_Duration_Share'
]

df_engineered[engineered_columns].head()
```

```python
# Reviewing the statistical summary of the engineered features.

df_engineered[engineered_columns].describe()
```

#### Handling Zero-Activity Sessions

```python
# Handling zero-activity sessions in the engineered ratio features.

ratio_columns = [
    'Product_Page_Share',
    'Avg_Duration_Per_Page',
    'Product_Duration_Share'
]

df_engineered[ratio_columns] = (
    df_engineered[ratio_columns].fillna(0)
)
```

```python
# Checking for missing values in the engineered features.

df_engineered[engineered_columns].isnull().sum()
```

#### Final Feature Engineering Validation

```python
# Checking the final dataset information after feature engineering.

df_engineered.info()
```

```python
# Saving the feature-engineered dataset for the next stage of the project.

df_engineered.to_csv(
    '../data/processed/feature_engineered_data.csv',
    index=False
)
```
