#### Importing the libraries

```python
#### Importing the libraries

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

#### Loading the dataset

```python
#### Loading the dataset

df = pd.read_csv('../data/raw/online_shoppers_intention.csv')
```

#### Univariate Analysis
Numerical Variables

```python
#### Distribution of Page Visits

page_visit_columns = [
    'Administrative',
    'Informational',
    'ProductRelated'
]

for col in page_visit_columns:

    fig = px.histogram(
        df,
        x=col,
        nbins=30,
        title=f'Distribution of {col} Page Visits',
        labels={
            col: 'Number of Pages Visited',
            'count': 'Number of Sessions'
        }
    )

    fig.update_layout(
        template='plotly_white',
        xaxis_title='Number of Pages Visited',
        yaxis_title='Number of Sessions'
    )

    fig.show()
```

```python
#### Distribution of Page Category Durations

duration_columns = [
    'Administrative_Duration',
    'Informational_Duration',
    'ProductRelated_Duration'
]

for col in duration_columns:

    fig = px.histogram(
        df,
        x=col,
        nbins=30,
        title=f'Distribution of {col}',
        labels={
            col: 'Duration',
            'count': 'Number of Sessions'
        }
    )

    fig.update_layout(
        template='plotly_white',
        xaxis_title='Duration',
        yaxis_title='Number of Sessions'
    )

    fig.show()
```

```python
#### Distribution of Browsing Behavior Metrics

behavior_columns = [
    'BounceRates',
    'ExitRates',
    'PageValues'
]

for col in behavior_columns:

    fig = px.histogram(
        df,
        x=col,
        nbins=30,
        title=f'Distribution of {col}',
        labels={
            col: col,
            'count': 'Number of Sessions'
        }
    )

    fig.update_layout(
        template='plotly_white',
        xaxis_title=col,
        yaxis_title='Number of Sessions'
    )

    fig.show()
```

Numerical Feature Correlation Heatmap

```python
#### Correlation Analysis

correlation_columns = [
    'Administrative',
    'Administrative_Duration',
    'Informational',
    'Informational_Duration',
    'ProductRelated',
    'ProductRelated_Duration',
    'BounceRates',
    'ExitRates',
    'PageValues',
    'SpecialDay',
    'Revenue'
]

correlation_df = df[correlation_columns].copy()

# Convert Revenue from boolean to binary numeric values
correlation_df['Revenue'] = correlation_df['Revenue'].astype(int)

correlation_matrix = correlation_df.corr()

fig = px.imshow(
    correlation_matrix,
    text_auto='.2f',
    aspect='auto',
    title='Numerical Features Show Varying Relationships with Purchase Conversion',
    labels={
        'x': 'Features',
        'y': 'Features',
        'color': 'Correlation'
    }
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Features',
    yaxis_title='Features'
)

fig.show()
```
