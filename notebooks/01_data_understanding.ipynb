#### Based on the dataset documentation, the meaning of each column in the Online Shoppers Purchasing Intention dataset is as follows:

* **Administrative** - Number of administrative pages visited by the visitor during the session.
* **Administrative_Duration** - Total time spent by the visitor on administrative pages during the session.
* **Informational** - Number of informational pages visited by the visitor during the session.
* **Informational_Duration** - Total time spent by the visitor on informational pages during the session.
* **ProductRelated** - Number of product-related pages visited by the visitor during the session.
* **ProductRelated_Duration** - Total time spent by the visitor on product-related pages during the session.
* **BounceRates** - Average bounce rate of the pages visited during the session.
* **ExitRates** - Average exit rate of the pages visited during the session.
* **PageValues** - Average value of pages visited by the visitor before completing a transaction.
* **SpecialDay** - Closeness of the visit to a special day or shopping occasion, such as Valentine’s Day or Mother’s Day.
* **Month** - Month in which the shopping session occurred.
* **OperatingSystems** - Operating system used by the visitor during the session.
* **Browser** - Browser used by the visitor during the session.
* **Region** - Geographic region associated with the visitor.
* **TrafficType** - Type of traffic source through which the visitor reached the website.
* **VisitorType** - Type of visitor, such as returning visitor or new visitor.
* **Weekend** - Indicates whether the shopping session occurred on a weekend.
* **Revenue** - Indicates whether the shopping session resulted in a purchase, serving as the primary target variable for conversion classification.

#### Checking unique values for each column

```python
for col in df.columns:
    unique_vals = df[col].unique()

    print(f'Column Name : {col}')
    print(f'Number of Unique Values : {len(unique_vals)}')
    print(f'Unique Values : {unique_vals}')
    print('-' * 50)
```

#### Checking the target variable distribution

```python
print("Revenue Value Counts:")
print(df['Revenue'].value_counts())

print("\nRevenue Percentage:")
print(df['Revenue'].value_counts(normalize=True) * 100)
```

#### Checking missing values per column, including empty strings

```python
# Count of NaN / None values per column
nan_count = df.isnull().sum()

# Count of empty strings per column
empty_count = (df == '').sum()

# Total missing values
total_missing = nan_count + empty_count

# Percentage of missing values
missing_percentage = (total_missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': total_missing,
    'Percentage (%)': missing_percentage
}, index=df.columns).sort_values(
    by='Percentage (%)',
    ascending=False
)

print(missing_df)
```
