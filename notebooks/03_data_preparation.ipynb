#### Importing Libraries

```python
#### Importing the libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

#### Loading the Dataset

```python
#### Loading the dataset

df = pd.read_csv('../data/raw/online_shoppers_intention.csv')

df.head()
```

```python
# Creating a copy of the original dataset for data preparation.

df_prepared = df.copy()
```

#### Checking the Initial Data Types

```python
# Checking the initial data types of the dataset.

df_prepared.dtypes
```

#### Converting Categorical Columns

```python
# Defining the categorical columns.

categorical_columns = [
    'Month',
    'OperatingSystems',
    'Browser',
    'Region',
    'TrafficType',
    'VisitorType'
]

# Converting the categorical columns to category data type.

for column in categorical_columns:
    
    df_prepared[column] = (
        df_prepared[column].astype('category')
    )
```

```python
# Checking the updated data types after conversion.

df_prepared.dtypes
```

#### Checking for Duplicate Records

```python
# Even though the dataset may not contain duplicates, double checking to validate this before modeling.

# Checking the number of duplicate rows in the dataset.

duplicate_count = df_prepared.duplicated().sum()

print(
    "Number of duplicate rows:",
    duplicate_count
)
```

#### Removing Duplicate Records

```python
# Removing duplicate records from the prepared dataset.

df_prepared = df_prepared.drop_duplicates().reset_index(
    drop=True
)
```

```python
# Checking the number of duplicate rows after removal.

duplicate_count = df_prepared.duplicated().sum()

print(
    "Number of duplicate rows:",
    duplicate_count
)
```

#### Validating Missing Values

```python
# Checking for missing values before preparing the dataset.

missing_values = df_prepared.isnull().sum()

print(missing_values)
```

#### Checking for Invalid Negative Values

```python
# Defining the numerical columns for validation.

numerical_columns = [
    'Administrative',
    'Administrative_Duration',
    'Informational',
    'Informational_Duration',
    'ProductRelated',
    'ProductRelated_Duration',
    'BounceRates',
    'ExitRates',
    'PageValues',
    'SpecialDay'
]


# Checking for negative values in the numerical columns.

for column in numerical_columns:
    
    negative_values = (
        df_prepared[column] < 0
    ).sum()
    
    print(
        f"{column}: {negative_values} negative values"
    )
```

#### Validating Binary Columns

```python
# Defining the binary columns for validation.

binary_columns = [
    'Weekend',
    'Revenue'
]

# Checking the unique values in the binary columns.

for column in binary_columns:
    
    print(f"\n{column}:")
    print(df_prepared[column].unique())
```
